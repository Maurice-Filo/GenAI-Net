```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
```

```python
t = np.linspace(0, 100, 1000)  # Time vector
y_list = list([np.sin(2 * np.pi * f0 * t) * np.exp(-sigma * t).reshape(1, -1)  for f0, sigma in [(0.1, 0.01), (0.2, 0.02), (0.3, 0.03)]])  # Example signal
peaks_indices = [find_peaks(np.squeeze(y), height=0)[0] for y in y_list]
for peak_idx, y in zip(peaks_indices, y_list):
    plt.plot(t, np.squeeze(y))
    plt.plot(t[peak_idx], y[0,peak_idx], "x")
plt.xlabel('Time')
plt.ylabel('Output Signal')
plt.title('Signal with Peaks Marked')
plt.show()
```

```python
from RL4CRN.utils.utils import oscillation_metric
f_list = [0.05, 0.15, 0.25]  # Desired frequencies
t0 = 20.0  # Time after which to start considering peaks
metric = oscillation_metric(f_list, y_list, t, t0, 0.1)
print(f"Oscillation Metric: {metric}")
```
