```python
# Import packages
from RL4CRN.iocrns.reaction_library import construct_mass_action_library
from RL4CRN.iocrns.reactions import MassAction
from RL4CRN.iocrns.reaction_library import ReactionLibrary
```

```python
# Construct library from a list of reactions
rxn_0 = MassAction(reactant_labels=[], product_labels=['X_1'], input_channels=[None], params=[None], params_controllability=[False])
rxn_1 = MassAction(reactant_labels=['X_1'], product_labels=[], input_channels=[None], params=[None], params_controllability=[False])
rxn_2 = MassAction(reactant_labels=['X_1', 'X_1'], product_labels=['X_2', 'X_3', 'X_4'], input_channels=[None], params=[None], params_controllability=[False])
rxn_3 = MassAction(reactant_labels=['X_1', 'X_1'], product_labels=['X_2'], input_channels=[None], params=[None], params_controllability=[False])
reaction_library_0 = ReactionLibrary([rxn_0, rxn_1, rxn_2, rxn_3])

# Print the library
print(reaction_library_0)

# Print lookup tables
print(f"Parameter lookup table: (size = {len(reaction_library_0.parameter_lookup_table)})")
print(reaction_library_0.parameter_lookup_table)
print(f"Controllable parameter lookup table: (size = {len(reaction_library_0.controllable_parameter_lookup_table)})")
print(reaction_library_0.controllable_parameter_lookup_table)
```

```python
# Construct and print the parameter masks from the library
masks = {}
masks['continuous'] = reaction_library_0.get_parameter_mask(mode='continuous')
masks['discrete'] = reaction_library_0.get_parameter_mask(mode='discrete')
masks['logit'] = reaction_library_0.get_logit_mask()
print("Parameter masks:")
print(masks)
```

```python
# Construct a mass-action reaction library for a given set of species and reaction order
species_labels = ['X_1', 'X_2', 'X_3']
reaction_library = construct_mass_action_library(species_labels, order=2)
print(reaction_library)
```

```python
# Get reaction from ID
print(reaction_library.get_reaction(0))
print(reaction_library.get_reaction(87))

# Print specific reactions
reaction_library.print_reactions([0, 87])

# Print parameter lookup tables
print("Parameter lookup table:")
print(reaction_library.parameter_lookup_table)
print("Controllable parameter lookup table:")
print(reaction_library.controllable_parameter_lookup_table)
```

```python
# Construct and print the parameter masks from the library
masks = {}
masks['continuous'] = reaction_library.get_parameter_mask(mode='continuous')
masks['discrete'] = reaction_library.get_parameter_mask(mode='discrete')
masks['logit'] = reaction_library.get_logit_mask()
print("Parameter masks:")
print(masks)
```

```python
# Add reactions to the library
r_1 = MassAction(reactant_labels=['X_3'], product_labels=['X_4'], input_channels=[None], params=[None], params_controllability=[False])
r_2 = MassAction(reactant_labels=['X_1', 'X_2'], product_labels=['X_4'], input_channels=[None], params=[None], params_controllability=[False])
reaction_library.add_reactions([r_1, r_2])
reaction_library.prepare_lookup_tables()  # Re-prepare lookup tables after adding new reactions
print(reaction_library)

# Print the added reactions
reaction_library.print_reactions([91, 92])

# Print parameter lookup tables after adding new reactions
print("Parameter lookup table after adding new reactions:")
print(reaction_library.parameter_lookup_table)
print("Controllable parameter lookup table after adding new reactions:")
print(reaction_library.controllable_parameter_lookup_table)
```

```python
# Get the number of reactions, parameters and controllable parameters in the library
print(f'Number of reactions in the library: {len(reaction_library)}')
print(f'Number of parameters in the library: {reaction_library.get_num_parameters()}')
print(f'Number of controllable parameters in the library: {reaction_library.get_num_controllable_parameters()}')
```
