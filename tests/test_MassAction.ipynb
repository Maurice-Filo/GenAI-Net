```python
# Import packages
from RL4CRN.iocrns.reactions import MassAction
```

```python
# Constructing reactions
rxn_0 = MassAction(reactant_labels=[], product_labels=['X_1'], input_channels=[None], params=[None], params_controllability=[True])
rxn_1 = MassAction(reactant_labels=['X_1'], product_labels=[], input_channels=[None], params=[None], params_controllability=[False])
rxn_2 = MassAction(reactant_labels=['X_1', 'X_1'], product_labels=['X_2', 'X_3', 'X_4'], input_channels=['u_2'], params=[1], params_controllability=[False])
rxn_3 = MassAction(reactant_labels=['X_1', 'X_1'], product_labels=['X_2'], input_channels=['u_1'], params=[1.5], params_controllability=[True])
```

```python
# Print reactions
print(rxn_0)
print(rxn_1)
print(rxn_2)
print(rxn_3)
```

```python
# Get involved species and inputs
print("Involved species in reaction 0:", rxn_0.get_involved_species())
print("Involved species in reaction 1:", rxn_1.get_involved_species())
print("Involved species in reaction 2:", rxn_2.get_involved_species())
print("Involved species in reaction 3:", rxn_3.get_involved_species())
print("--------------------------------")
print("Involved inp uts in reaction 0:", rxn_0.get_involved_inputs())
print("Involved inputs in reaction 1:", rxn_1.get_involved_inputs())
print("Involved inputs in reaction 2:", rxn_2.get_involved_inputs())
print("Involved inputs in reaction 3:", rxn_3.get_involved_inputs())
```

```python
# Get stoichiometry
print("Stoichiometry of reaction 0:")
print(rxn_0.get_stoichiometry_dict())
print("Stoichiometry of reaction 1:")
print(rxn_1.get_stoichiometry_dict())
print("Stoichiometry of reaction 2:")
print(rxn_2.get_stoichiometry_dict())
print("Stoichiometry of reaction 3:")
print(rxn_3.get_stoichiometry_dict())
```
