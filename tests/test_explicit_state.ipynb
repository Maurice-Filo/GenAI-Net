```python
# Importing packages
import numpy as np
from itertools import product
from RL4CRN.iocrns.iocrn import IOCRN
from RL4CRN.iocrns.reactions import MassAction
from RL4CRN.iocrns.reaction_library import construct_mass_action_library
from RL4CRN.states.explicit_state_representation import ExplicitStateRepresentation
```

```python
# Create mass-action reactions
r1 = MassAction([], ['Z_1'], ['u_1'], [10.])
r2 = MassAction(['X_2'], ['X_2', 'Z_2'], [None], [1])
r3 = MassAction(['Z_1', 'Z_2'], [], [None], [0.1])
r4 = MassAction(['Z_1'], ['Z_1', 'X_1'], [None], [0.1])
r5 = MassAction(['X_1'], ['X_1', 'X_2'], [None], [0.1])
r6 = MassAction(['X_1'], [], ['u_3'], [1.])
r7 = MassAction(['X_2'], [], ['u_2'], [1.])
r8 = MassAction(['Z_1'], [], [None], [0.01])
r9 = MassAction(['Z_2'], [], [None], [0.01])
```

```python
# Create an IOCRN
iocrn = IOCRN(reactions=[r1, r2, r3, r4, r5, r6, r7, r8, r9], output_labels=['X_2'])
iocrn.compile()
iocrn.retrieve_species()
iocrn.retrieve_input()
print(iocrn)
```

```python
# Construct the reaction library
species_labels = ['X_1', 'X_2', 'Z_1', 'Z_2']
reaction_library = construct_mass_action_library(species_labels, order=2)
print(reaction_library)
```

```python
# Set the context of the IOCRN with respect to the reaction library
iocrn.set_library_context(reaction_library)
```

```python
# Create an ExplicitState
state = ExplicitStateRepresentation(iocrn, reaction_library)
structure, parameters, inputs = state()
print('Structure hot encoding, Type: ', type(structure), 'Shape: ', structure.shape, ':\n', structure, '\n')
print('Parameters encoding, Type: ', type(parameters), 'Shape: ', parameters.shape, '\n', parameters, '\n')
print('Inputs encoding, Type: ', type(inputs), 'Shape: ', inputs.shape, '\n', inputs, '\n')
```

```python
# Verify
print(iocrn)

# Structure verification
print('Structure:')
hot_indexes = np.argwhere(structure).flatten().tolist()
print(hot_indexes)
reaction_library.print_reactions(hot_indexes)

# Parameters verification
print('Parameters')
param_idx = np.flatnonzero(parameters).tolist()
print(parameters[param_idx])

# Inputs verification
print('Inputs')
input_idx_flat = np.argwhere(inputs).flatten().tolist()
input_idx = np.zeros(iocrn.num_inputs, dtype=int)
for i in range(iocrn.num_inputs):
    input_idx[i] = input_idx_flat[i] - i * reaction_library.get_num_controllable_parameters()
    
print(input_idx)
```
