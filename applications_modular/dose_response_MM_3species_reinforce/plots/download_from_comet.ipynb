```python
# Import packages
from comet_ml import API
import os
```

```python
# API and experiment key
API_KEY = "o77J6VCMDamustkfJuMXZ2jdV"
EXPERIMENT_KEY = "a7e4dc88100a4ee8a75320f73dd0e565"

# Selet asset type and name
asset_type = 'image'
asset_name_starting = 'CRN Distribution'

# Output directory
out_dir = "trajectory_distributions"

# Base name for saved files
base_name = 'dose_response_3species_5rxns_MM_REINFORCE'
```

```python
# Set up API, experiment and create output directory
api = API(api_key=API_KEY)
exp = api.get_experiment_by_key(EXPERIMENT_KEY)
print(exp.url)
os.makedirs(out_dir, exist_ok=True)

# Get all assets
assets = exp.get_asset_list()
```

```python
# Keep only selected asset type
selected_assets = [a for a in assets if a['type'] == (asset_type)]
print(f"Found {len(selected_assets)} assets of type '{asset_type}'")

# Sort images by step
sorted_selected_assets = sorted(selected_assets, key=lambda x: x['step'])
print()
```

```python
# Keep only the assets with the selected name
final_assets = [a for a in sorted_selected_assets if a['fileName'].startswith(asset_name_starting)]
```

```python
# Loop over and download each asset
for asset in final_assets:
    asset_id = asset["assetId"]
    step = asset.get("step", -1)
    data = exp.get_asset(asset_id)

    # Build a readable filename: e.g. "CRN_Distribution_step_000042_ab12cd34.png"
    filename = f"{base_name}_{int(step)}.png"
    file_path = os.path.join(out_dir, filename)

    with open(file_path, "wb") as f:
        f.write(data)

    print(f"Saved {file_path}")

print(f"\nDownloaded {len(final_assets)} image(s) to '{out_dir}'")
```
