```python
# Import packages
from PIL import Image
import os, re
```

```python
# Set input directory and output path
in_dir = "trajectory_distributions"
out_path = "gifs/dose_response_3species_5rxns_MM_REINFORCE.gif"
downsample_factor = 4
```

```python
# Collect and sort PNGs
files = [f for f in os.listdir(in_dir)]

files = sorted(files, key=lambda x: int(x.split("_")[-1].split(".")[0]))

# Load frames
frames = [Image.open(os.path.join(in_dir, f)) for f in files]
```

```python
# Save as GIF
frames[0].save(
    out_path,
    save_all=True,
    append_images=frames[0::downsample_factor],
    duration=1, # ms per frame (e.g., 100 = 10 fps)
    loop=None,
    optimize=True,
)

print(f"Saved {out_path} with {len(frames)} frames")
```
