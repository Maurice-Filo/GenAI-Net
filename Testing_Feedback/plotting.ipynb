```python
import networkx as nx
import matplotlib.pyplot as plt

# Create directed graph
G = nx.DiGraph()

# Nodes
species = ['X₁', 'Z₁', 'Z₂']
inputs = ['u₁', 'u₂']
reactions = [f'R{i}' for i in range(8)]

# Add species and input nodes
G.add_nodes_from(species, shape='s')
G.add_nodes_from(inputs, shape='o')
G.add_nodes_from(reactions, shape='circle')

# Reactions (edges)
edges = [
    ('u₁', 'R0'), ('R0', 'Z₁'),  # Reaction 0
    ('X₁', 'R1'), ('R1', '∅'),  # Reaction 1
    ('X₁', 'R2'), ('Z₂', 'R2'), ('R2', 'Z₂'),  # Reaction 2
    ('X₁', 'R3'), ('X₁', 'R3'), ('R3', '∅'),  # Reaction 3
    ('X₁', 'R4'), ('Z₂', 'R4'), ('R4', '∅'),  # Reaction 4
    ('Z₁', 'R5'), ('R5', 'X₁'), ('R5', 'Z₂'),  # Reaction 5
    ('Z₁', 'R6'), ('Z₂', 'R6'), ('R6', 'Z₁'), ('R6', 'Z₁'),  # Reaction 6
    ('X₁', 'R7'), ('X₁', 'R7'), ('R7', 'X₁'),  # Reaction 7
]

# Add edges to graph
G.add_edges_from(edges)

# Plotting positions
pos = nx.spring_layout(G, seed=42)

# Draw nodes by type
nx.draw_networkx_nodes(G, pos, nodelist=species, node_shape='s', node_color='lightblue')
nx.draw_networkx_nodes(G, pos, nodelist=inputs, node_shape='o', node_color='lightgreen')
nx.draw_networkx_nodes(G, pos, nodelist=reactions, node_shape='o', node_color='orange')

# Draw edges and labels
nx.draw_networkx_edges(G, pos)
nx.draw_networkx_labels(G, pos)

# Rate labels manually positioned (for clarity)
rate_labels = {
    ('u₁', 'R0'): '1.0u₁',
    ('X₁', 'R1'): '1.0u₂',
    ('R2', 'Z₂'): '3.01',
    ('R3', '∅'): '0.89',
    ('R4', '∅'): '1.42',
    ('R5', 'X₁'): '2.74',
    ('R6', 'Z₁'): '3.46',
    ('R7', 'X₁'): '2.80',
}
nx.draw_networkx_edge_labels(G, pos, edge_labels=rate_labels)

# improve layout
plt.title('CRN 45_0')
plt.axis('equal')
plt.show()
```
