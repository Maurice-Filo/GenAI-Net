```python
# Import packages and modules
import numpy as np
from RL4CRN_Feedback.Input_Output_Rxn_Networks.IOCRN_MassAction import IOCRN_MassAction
```

```python
# Construct the basic CRN
species_labels = ['X_1', 'Z_1', 'Z_2']
inputs_labels = ['u_1', 'u_2']
stoichiometry_reactants = np.array([[0], [1], [0]], dtype=np.int8)
stoichiometry_products = np.array([[1], [1], [0]], dtype=np.int8)
k = 1
parameters = np.array([k], dtype=np.float32)
input_influence_matrix = np.array([[0], [0]], dtype=np.int8)
outputs = np.array([1], dtype=np.int8)
IOCRN_0 = IOCRN_MassAction(stoichiometry_reactants, stoichiometry_products, parameters, input_influence_matrix, outputs, species_labels, inputs_labels)
```

```python
# Print all possible complexex 
print(f"Number of Complexes: {IOCRN_0.get_complexes_range() - 1}")
for i in range(IOCRN_0.get_complexes_range()):
    species_1, species_2 = IOCRN_0.map_complex_to_species(i)
    print(f"C_{i} = X_{species_1} + X_{species_2}")
```

```python
# Print all possible reactions
print(f"Number of Reactions: {IOCRN_0.get_reactions_range() - 1}")
print("Reaction r: (i, j, k, l) means R_r: X_i + X_j -> X_k + X_l")
for i in range(IOCRN_0.get_reactions_range()):
    reactant_1, reactant_2, product_1, product_2 = IOCRN_0.map_reaction_to_species(i)
    reactants, products = IOCRN_0.map_reaction_to_complexes(i)
    print(f"R_{i}: X_{reactant_1} + X_{reactant_2} -> X_{product_1} + X_{product_2}, C_{reactants} -> C_{products}")
```
