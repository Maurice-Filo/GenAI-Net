```python
# Import packages
from PIL import Image
import os, re
```

```python
# Set input directory and output path
in_dir1 = "trajectory_distributions"
in_dir2 = "frequency_content_distributions"
out_path = "gifs/freq_oscillators_3species_6rxns_MAK_REINFORCE.gif"
downsample_factor = 1
```

```python
# Collect and sort PNGs
files1 = [f for f in os.listdir(in_dir1)]
files2 = [f for f in os.listdir(in_dir2)]

files1 = sorted(files1, key=lambda x: int(x.split("_")[-1].split(".")[0]))
files2 = sorted(files2, key=lambda x: int(x.split("_")[-1].split(".")[0]))

# Ensure both directories have the same number of frames
num_frames = min(len(files1), len(files2))

# Load frames
frames = []
for i in range(num_frames):
    img1 = Image.open(os.path.join(in_dir1, files1[i]))
    img2 = Image.open(os.path.join(in_dir2, files2[i]))

    # Ensure both images are the same height
    if img1.height != img2.height:
        # Resize img2 to match img1 height while preserving aspect ratio
        new_width = int(img2.width * img1.height / img2.height)
        img2 = img2.resize((new_width, img1.height))

    # Create new combined image (side by side)
    combined_width = img1.width + img2.width
    combined = Image.new("RGB", (combined_width, img1.height))
    combined.paste(img1, (0, 0))
    combined.paste(img2, (img1.width, 0))

    frames.append(combined)
```

```python
# Save as GIF
frames[0].save(
    out_path,
    save_all=True,
    append_images=frames[0::downsample_factor],
    duration=1, # ms per frame (e.g., 100 = 10 fps)
    loop=None,
    optimize=True,
)

print(f"Saved {out_path} with {len(frames)} frames")
```
