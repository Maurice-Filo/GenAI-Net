```python
# Import packages
import json
import matplotlib.pyplot as plt
```

```python
# Filenames
n = 10
filenames = [f"losses_files/average loss in the batch top k,best loss in the batch top k,worst loss in the batch top k VS step_chart_data_{i}.json" for i in range(n)]
saved_filenames = [f"losses_plots/attempt_{i}.png" for i in range(n)]
save_flag = True

for filename, saved_filename in zip(filenames, saved_filenames):
    with open(filename, "r") as f:
        data = json.load(f)

    for trace in data:
        plt.plot(trace["x"], trace["y"], label=trace["name"])

    plt.xlim(0, 500)
    plt.ylim(0, 0.5)
    plt.xlabel("Step")
    plt.ylabel("Loss")
    plt.title("Batch Top-K Losses")
    plt.legend()
    if save_flag:
        plt.savefig(saved_filename)
    plt.show()
    
```
